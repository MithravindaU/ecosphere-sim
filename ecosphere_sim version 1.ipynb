```python
import numpy as np
import matplotlib.pyplot as plt
import random
import time
from IPython.display import clear_output

world = np.zeros((20, 20))
```

```python

for i in range(50):

    x = np.random.randint(0, 20)

    y = np.random.randint(0, 20)

    world[x][y] = 1
```

```python
herbivores = []

for i in range(5):

    x = np.random.randint(0,20)

    y = np.random.randint(0,20)

    herbivores.append([x,y])
```

```python
for step in range(100):

  display_world = world.copy()
```

```python
for animal in herbivores:

  dx = random.choice(
    [-1, 0, 1]
)

dy = random.choice(
    [-1, 0, 1]
)
animal[0] += dx
animal[1] += dy

animal[0] = max(
    0,
    min(19, animal[0])
)

animal[1] = max(
    0,
    min(19, animal[1])
)
display_world[
            animal[0]
        ][
            animal[1]
        ] = 2
```

```python
clear_output(wait=True)

plt.figure(figsize=(6,6))

plt.imshow(display_world)

plt.title(
        f"Step: {step}"
    )

plt.show()


time.sleep(0.2)
```
